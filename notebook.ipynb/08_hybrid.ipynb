```python
import pandas as pd
import numpy as np

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error

import matplotlib.pyplot as plt

import mlflow
```

```python
prophet = pd.read_csv("../Data/prophet_forecast.csv")

prophet.head()
```

```python
lstm = pd.read_csv("../Data/lstm_forecast.csv")

lstm.head()
```

```python
lstm.columns = [
    "Day",
    "LSTM_Prediction"
]
```

```python
lstm.rename(
    columns={
        "Prediction":"LSTM_Prediction"
    },
    inplace=True
)
```

```python
hybrid = prophet.tail(len(lstm)).copy()

hybrid.reset_index(
    drop=True,
    inplace=True
)

lstm.reset_index(
    drop=True,
    inplace=True
)

hybrid["LSTM"] = lstm["LSTM_Prediction"]
```

```python
hybrid["HybridForecast"] = (

    hybrid["yhat"] +

    hybrid["LSTM"]

)/2
```

```python
hybrid.to_csv(

    "hybrid_forecast.csv",

    index=False

)

print("Hybrid Forecast Saved")
```
