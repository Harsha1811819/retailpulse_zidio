## 1. Customer Segmentation

We generate mock RFM (Recency, Frequency, Monetary) data for 300 customers, perform Principal Component Analysis (PCA) to reduce dimensions down to 2 components, and visualize the segments.

```python
import pandas as pd
import numpy as np
import plotly.express as px
from sklearn.decomposition import PCA

# 1. Load mock segments
np.random.seed(42)
n = 300
df_seg = pd.DataFrame({
    "Recency": np.random.randint(1, 200, n),
    "Frequency": np.random.randint(1, 50, n),
    "Monetary": np.random.randint(100, 5000, n),
    "Segment": np.random.choice(["Champions", "Loyal", "At Risk", "Potential", "Hibernating", "New"], n)
})

print("Total Customers:", len(df_seg))
print("Segments:", df_seg['Segment'].nunique())
df_seg.head()
```

```python
# 2. Apply PCA
features = df_seg[["Recency", "Frequency", "Monetary"]]
pca = PCA(n_components=2)
comps = pca.fit_transform(features)
df_seg["PCA1"] = comps[:, 0]
df_seg["PCA2"] = comps[:, 1]

print("Explained variance ratio:", pca.explained_variance_ratio_)
df_seg.head()
```

```python
# 3. Plot PCA Projections
fig_seg = px.scatter(
    df_seg,
    x="PCA1",
    y="PCA2",
    color="Segment",
    title="Customer Segments (PCA Projection)",
    hover_data=["Recency", "Frequency", "Monetary"]
)
fig_seg.show()
```

```python
# 4. Segment KPI Summary
summary = df_seg.groupby("Segment").agg({
    "Recency": "mean",
    "Frequency": "mean",
    "Monetary": "mean"
}).round(2)
summary
```

## 2. Churn Risk Analytics

We generate mock churn risk predictions for 20 customers, analyze their churn probabilities, filter top at-risk customers, and visualize the risk distributions.

```python
# 1. Load mock churn prediction data
np.random.seed(42)
customers = []
reasons = ["High Recency", "Low Frequency", "No Recent Orders", "Low Spend", "Declining Activity"]

for i in range(1, 21):
    customers.append({
        "CustomerID": ff"CUST{i:04d}",
        "ChurnProbability": round(np.random.uniform(0.70, 0.99), 2),
        "TopReason1": np.random.choice(reasons),
        "TopReason2": np.random.choice(reasons),
        "TopReason3": np.random.choice(reasons)
    })

df_churn = pd.DataFrame(customers)
print("Customers Monitored:", len(df_churn))
print("Average Risk:", f"{df_churn['ChurnProbability'].mean():.2%}")
df_churn.head()
```

```python
# 2. Get Top At-Risk Customers (e.g. top 10)
top_n = 10
risk_df = df_churn.sort_values("ChurnProbability", ascending=False).head(top_n)
risk_df
```

```python
# 3. Plot Top Churn Risks
fig_churn = px.bar(
    risk_df,
    x="CustomerID",
    y="ChurnProbability",
    color="ChurnProbability",
    title="Top Customer Churn Risks"
)
fig_churn.show()
```
