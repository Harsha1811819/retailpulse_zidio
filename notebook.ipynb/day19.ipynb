# Day 19: Real-Time Metrics & Alerts

This notebook demonstrates real-time pipeline monitoring, statistical feature drift detection, and automated anomaly alerting frameworks.

```python
import pandas as pd
import numpy as np
import plotly.express as px

# 1. Load mock drift monitoring summary
features = ["Quantity", "Price", "Recency", "Frequency", "Monetary"]
p_values = [0.012, 0.043, 0.482, 0.812, 0.651]
drifted = [p < 0.05 for p in p_values]

df_drift = pd.DataFrame({
    "Feature": features,
    "p_value": p_values,
    "DriftDetected": drifted,
    "Status": ["Alert! Drift 🔴" if d else "Stable 🟢" for d in drifted],
    "Details": ["Artificial Drift Added" if d else "Stable Distribution" for d in drifted]
})

df_drift
```

```python
# 2. Plotting a Distribution Shift (Drift Visualizer)
np.random.seed(42)
n = 1000
baseline = np.random.normal(100, 15, n)
current_drifted = np.random.normal(108, 18, n)

df_dist = pd.DataFrame({
    "Value": np.concatenate([baseline, current_drifted]),
    "Dataset": ["Baseline (Training)"] * n + ["Current (Inference)"] * n
})

fig = px.histogram(
    df_dist,
    x="Value",
    color="Dataset",
    barmode="overlay",
    marginal="box",
    title="Distribution Shift for 'Quantity' (Drift Detected)",
    color_discrete_map={"Baseline (Training)": "#1f77b4", "Current (Inference)": "#ff7f0e"}
)
fig.show()
```

## Simulated Real-Time Streaming Alerts

We write a function that acts as a real-time monitor checking a stream of incoming transactions and raising alerts when quantity anomalies or value thresholds are breached.

```python
import time

def check_transaction_stream(stream_data):
    print("Starting Real-Time Transaction Stream Monitoring...")
    print("-" * 60)
    alerts_triggered = 0
    
    for idx, row in stream_data.iterrows():
        qty = row["Quantity"]
        price = row["Price"]
        total = qty * price
        
        # Check conditions
        if qty > 100:
            print(f"⚠️ ALERT [Anomaly Qty] Row {idx}: Extreme Quantity ordered ({qty} units)!")
            alerts_triggered += 1
        if total > 500.0:
            print(f"🚨 ALERT [High Value] Row {idx}: High value invoice transaction (${total:,.2f})!")
            alerts_triggered += 1
            
    print("-" * 60)
    print(f"Stream Monitoring Complete. Total Alerts Raised: {alerts_triggered}")

# Create incoming stream
np.random.seed(42)
stream_df = pd.DataFrame({
    "Quantity": [10, 5, 120, 2, 85, 150, 4],
    "Price": [12.5, 4.0, 5.0, 99.0, 3.5, 1.2, 250.0]
})

check_transaction_stream(stream_df)
```
