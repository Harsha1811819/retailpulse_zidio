# Day 20: Export Functionality (CSV & PDF Reports)

This notebook demonstrates how to programmatically compile and export report documents as standard CSV tables and premium, print-ready PDF files using `reportlab`.

```python
import pandas as pd
import numpy as np

# 1. Compile summary analytics report data
summary_data = pd.DataFrame({
    "MetricName": [
        "Hybrid Forecast MAPE",
        "Churn Prediction Model AUC",
        "Inventory Stockout Reduction",
        "Active Drift Alerts",
        "Total Monitored SKUs"
    ],
    "Value": ["4.61%", "0.94", "32.7%", "2", "10"],
    "Status": ["Target Achieved", "Optimized", "Projected", "Action Required", "Stable"]
})

summary_data
```

```python
# 2. Export summary table as CSV
csv_filename = "retailpulse_summary_report.csv"
summary_data.to_csv(csv_filename, index=False)
print(f"Successfully saved CSV report as: {csv_filename}")
```

## Generating a PDF Report via ReportLab

We use Python's `reportlab` package to build a PDF document containing structural tables, branding, and stylized headers.

```python
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib import colors

pdf_filename = "retailpulse_executive_report.pdf"

# Set up template
doc = SimpleDocTemplate(pdf_filename, pagesize=letter, rightMargin=50, leftMargin=50, topMargin=50, bottomMargin=50)
story = []

# Styles
styles = getSampleStyleSheet()
title_style = ParagraphStyle(
    'TitleStyle',
    parent=styles['Heading1'],
    fontSize=22,
    textColor=colors.HexColor('#1f77b4'),
    spaceAfter=15
)
body_style = styles['Normal']

# 1. Add Title & Branding
story.append(Paragraph("🛒 RetailPulse Analytics Platform", title_style))
story.append(Paragraph("Executive Performance Summary Report", styles['Heading2']))
story.append(Spacer(1, 15))
story.append(Paragraph("This report contains high-level KPIs and pipeline checks generated automatically by the platform's background scheduling engines.", body_style))
story.append(Spacer(1, 20))

# 2. Add Table Data
table_data = [['Metric Name', 'Value', 'Status']]
for idx, row in summary_data.iterrows():
    table_data.append([row['MetricName'], row['Value'], row['Status']])
    
report_table = Table(table_data, colWidths=[200, 100, 150])
report_table.setStyle(TableStyle([
    ('BACKGROUND', (0,0), (-1,0), colors.HexColor('#1f77b4')),
    ('TEXTCOLOR', (0,0), (-1,0), colors.whitesmoke),
    ('ALIGN', (0,0), (-1,-1), 'LEFT'),
    ('FONTNAME', (0,0), (-1,0), 'Helvetica-Bold'),
    ('FONTSIZE', (0,0), (-1,0), 11),
    ('BOTTOMPADDING', (0,0), (-1,0), 8),
    ('BACKGROUND', (0,1), (-1,-1), colors.HexColor('#f5f5f5')),
    ('TEXTCOLOR', (0,1), (-1,-1), colors.black),
    ('FONTNAME', (0,1), (-1,-1), 'Helvetica'),
    ('FONTSIZE', (0,1), (-1,-1), 10),
    ('GRID', (0,0), (-1,-1), 1, colors.HexColor('#cccccc')),
    ('TOPPADDING', (0,0), (-1,-1), 6),
    ('BOTTOMPADDING', (0,0), (-1,-1), 6),
]))

story.append(report_table)
story.append(Spacer(1, 30))
story.append(Paragraph("Report generated successfully. End of document.", styles['Italic']))

# Build PDF
doc.build(story)
print(f"Successfully saved executive PDF report as: {pdf_filename}")
```
